```python
import os
import urllib.request

emotion_music_urls = {
    "happy": [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-1.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-8.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-10.mp3",
    ],
    "sad": [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-2.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-14.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-16.mp3",
    ],
    "angry": [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-3.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-11.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-17.mp3",
    ],
    "fear": [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-4.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-13.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-15.mp3",
    ],
    "neutral": [
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-6.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-12.mp3",
        "https://www.soundhelix.com/examples/mp3/SoundHelix-Song-18.mp3",
    ]
}

base_path = "/content/emotion_music"

for emotion, urls in emotion_music_urls.items():
    folder_path = os.path.join(base_path, emotion)
    os.makedirs(folder_path, exist_ok=True)
    for i, url in enumerate(urls, start=1):
        filename = f"{emotion}{i}.mp3"
        filepath = os.path.join(folder_path, filename)
        if os.path.exists(filepath): continue  # ✅ Skip if already downloaded
        try:
            urllib.request.urlretrieve(url, filepath)
            print(f"✅ Downloaded {filename} → {emotion}")
        except Exception as e:
            print(f"❌ Failed to download {filename} → {emotion}. Error: {e}")
```

```python
# === Cell 1 ===
from moviepy.editor import VideoFileClip
from pydub import AudioSegment
import cv2
import os
import random
import urllib.request
import shutil
import gradio as gr
import moviepy.editor as mp
from deepface import DeepFace
import random
import subprocess
from IPython.display import Audio


# --- Extract audio from video ---
def extract_audio(video_path):
    video = VideoFileClip(video_path)
    audio_path = "/content/extracted_audio.wav"
    video.audio.write_audiofile(audio_path)
    return audio_path

# --- Simulate vocal separation (placeholder) ---
import subprocess
def separate_audio(audio_path):
    output_dir = "/content/output"
    subprocess.run(["spleeter", "separate", "-o", output_dir, audio_path])
    vocals_path = os.path.join(output_dir, "extracted_audio", "vocals.wav")
    print(f"Vocal separation completed. Vocals saved at: {vocals_path}")
    return vocals_path

# --- Extract frames from video every 5 seconds ---
def extract_frames(video_path, interval_sec=20):
    frames = []
    output_dir = "/content/frames"
    os.makedirs(output_dir, exist_ok=True)

    vidcap = cv2.VideoCapture(video_path)
    fps = int(vidcap.get(cv2.CAP_PROP_FPS))  # Get frames per second
    frame_interval = fps * interval_sec       # Calculate frame interval for 5 seconds

    success, image = vidcap.read()
    count = 0
    frame_index = 0
    while success:
        if count % frame_interval == 0:
            frame_path = os.path.join(output_dir, f"frame{frame_index}.jpg")
            cv2.imwrite(frame_path, image)
            frames.append(frame_path)
            print(f"✅ Frame {frame_index} extracted")
            frame_index += 1
        success, image = vidcap.read()
        count += 1

    vidcap.release()
    return frames

# --- Detect emotion using DeepFace ---
def detect_emotion_from_frames(frames_folder):
    emotion_counts = {}
    for frame in sorted(os.listdir(frames_folder)):
        frame_path = os.path.join(frames_folder, frame)
        try:
            analysis = DeepFace.analyze(frame_path, actions=['emotion'], enforce_detection=False)
            dominant_emotion = analysis[0]['dominant_emotion']
            emotion_counts[dominant_emotion] = emotion_counts.get(dominant_emotion, 0) + 1
            print(f"{frame} ➝ {dominant_emotion}")
        except:
            print(f"{frame} ➝ No face detected")
    # Return the most frequent detected emotion
    if emotion_counts:
        return max(emotion_counts, key=emotion_counts.get)
    else:
        return "neutral"

def loop_music_to_duration(audio_segment, duration_ms):
    """Loop the music until it reaches or exceeds the duration, then trim."""
    looped = AudioSegment.silent(duration=0)
    while len(looped) < duration_ms:
        looped += audio_segment
    return looped[:duration_ms]



# --- Emotion to music mapping with folder structure ---
def emotion_to_music(emotion):
    emotion_folder_map = {
        "happy": "happy",
        "surprise": "happy",
        "sad": "sad",
        "angry": "angry",
        "fear": "fear",
        "neutral": "neutral",
        "calm": "neutral"
    }

    # Default to 'neutral' if unknown emotion
    folder_name = emotion_folder_map.get(emotion, "neutral")
    emotion_music_dir = os.path.join("/content/emotion_music", folder_name)

    if not os.path.exists(emotion_music_dir):
        print(f"⚠️ Folder not found for emotion: {emotion} ➝ Using fallback from /neutral")
        emotion_music_dir = os.path.join("/content/emotion_music", "neutral")

    music_files = [os.path.join(emotion_music_dir, f)
                  for f in os.listdir(emotion_music_dir)
                  if f.endswith(".mp3")]

    if not music_files:
        print(f"⚠️ No music files found in: {emotion_music_dir} ➝ Using hardcoded fallback track")
        return "/content/emotion_music/neutral/neutral1.mp3"  # Adjust this path to an existing file

    selected_track = random.choice(music_files)
    print(f"🎵 Random music selected for {emotion}: {selected_track}")
    return selected_track

```

```python
# === Cell 2 ===

# --- Full video processing pipeline ---
def run_full_pipeline(video_path):
    # Step 1: Extract & separate audio
    audio_path = extract_audio(video_path)
    vocals_path = separate_audio(audio_path)

    # Step 2: Extract frames and detect emotion
    frames = extract_frames(video_path)
    frames_folder = "/content/frames"
    emotion = detect_emotion_from_frames(frames_folder)
    print(f"✅ Detected Emotion: {emotion}")

    # Step 3: Load video to get duration
    video = mp.VideoFileClip(video_path)
    video_duration_ms = int(video.duration * 1000)

    # Step 4: Select background music
    music_file = emotion_to_music(emotion)
    print(f"🎵 Selected Music: {music_file}")
    background = AudioSegment.from_file(f"{music_file}")
   #background = background[:video_duration_ms]
    background = loop_music_to_duration(background, video_duration_ms)

    # Step 5: Load vocals, adjust and combine
    vocals = AudioSegment.from_file(vocals_path)
    vocals = vocals[:video_duration_ms]
    background_lowered = background - 10
    combined = background_lowered.overlay(vocals)
    combined = combined[:video_duration_ms]

    # Step 6: Export final audio
    combined_audio_path = "/content/final_combined_audio.mp3"
    combined.export(combined_audio_path, format="mp3")
    print("✅ Final audio exported.")

    # Step 7: Merge final audio with video
    output_video_path = "/content/final_video_with_music.mp4"
    final_audio = mp.AudioFileClip(combined_audio_path)
    final_video = video.set_audio(final_audio)
    final_video.write_videofile(output_video_path, codec="libx264", audio_codec="aac")
    return output_video_path

# --- Gradio Integration ---
def process_video(video_file):
    input_path = "/content/input_video.mp4"
    shutil.copy(video_file, input_path)
    return run_full_pipeline(input_path)

interface = gr.Interface(
    fn=process_video,
    inputs=gr.Video(label="Upload a Video"),
    outputs=gr.Video(label="Final Video with Emotion-Based Music"),
    title="AI-Based Music Composition Using Transformer Network",
    description="Upload a video, The system detects facial emotions and overlays emotion-matching background music.",
)

interface.launch(share=True, debug=True)
```
